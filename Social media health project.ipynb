# 10. Correlation Heatmap

# 📊 Analysis

#### The correlation heatmap visualizes the relationships between all numerical variables in the dataset. Correlation coefficients range from -1 to +1, where:

- #### +1 indicates a strong positive correlation.
- #### -1 indicates a strong negative correlation.
- #### 0 indicates little or no linear relationship.

#### The color gradient (red to blue) helps identify the strength and direction of the relationships between variables. Values displayed inside each cell provide the exact correlation coefficient, making the interpretation more accurate.This visualization is useful for understanding how factors such as Average Daily Social Media Usage, Sleep Duration, Academic Performance, Mental Health Score, and Age are related.

```python
import plotly.graph_objects as go

corr = df.select_dtypes(include="number").corr()

fig = go.Figure(
    data=go.Heatmap(
        z=corr.values,
        x=corr.columns,
        y=corr.columns,
        colorscale="RdBu",
        text=corr.round(2),
        texttemplate="%{text}",
        hoverongaps=False
    )
)

fig.update_layout(
    title="Correlation Heatmap",
    title_x=0.5,
    template="plotly_dark",
    width=800,
    height=700
)

fig.show()
 
```

- Academic Performance and Mental Health Score demonstrate a relationship, indicating that student well-being and academic outcomes are associated.
- Most correlations are weak to moderate, suggesting that no single factor completely explains the others.
- The diagonal values are 1.00, representing the perfect correlation of each variable with itself.
- The heatmap quickly highlights the strongest positive and negative relationships, helping identify variables that deserve further investigation.

#### Research Question 4. Does excessive social media usage influence academic performance?

```python
import pandas as pd
import plotly.express as px

# Load dataset
df = pd.read_csv("Social_Media_Health_Cleaned.csv")

# Create Usage Groups
df["Usage_Group"] = pd.cut(
    df["Avg_Daily_Usage_Hours"],
    bins=[0, 3, 6, 9],
    labels=["Low", "Moderate", "High"]
)

# Count students
academic = (
    df.groupby(["Usage_Group", "Affects_Academic_Performance"])
      .size()
      .reset_index(name="Count")
)

# Plotly Stacked Bar Chart
fig = px.bar(
    academic,
    x="Usage_Group",
    y="Count",
    color="Affects_Academic_Performance",
    barmode="stack",
    text="Count",
    title="Impact of Social Media Usage on Academic Performance",
    template="plotly_dark",
    color_discrete_sequence=px.colors.qualitative.Set2
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="Social Media Usage Level",
    yaxis_title="Number of Students",
    legend_title="Academic Performance"
)

fig.show()
```

# 📊 Analysis

#### The stacked bar chart categorizes students into Low (0–3 hours), Moderate (3–6 hours), and High (6–9 hours) social media usage groups and compares whether they reported that social media affects their academic performance.

# 🔍 Insights
1. Moderate usage group contains the largest number of students, indicating that most students spend between 3 and 6 hours per day on social media.
2. Students in the High usage group are more likely to report that social media affects their academic performance, suggesting that excessive usage       may interfere with concentration and study time.
3. The Low usage group has the fewest students reporting a negative academic impact, indicating that limited social media usage is generally              associated with fewer academic concerns.
4. As daily social media usage increases, the proportion of students reporting an impact on academic performance also increases, showing a clear          upward trend.
5. The comparison between "Yes" and "No" responses across usage groups suggests a relationship between usage duration and academic outcomes, although     the chart alone does not establish causation.
6. The stacked bar chart effectively highlights differences among usage levels, making it easy to compare the distribution of students affected and       not affected within each category.
7. Students with moderate usage may still maintain academic performance, but continued increases in usage appear to be associated with greater            academic challenges.
8. These findings emphasize the importance of balanced social media usage, especially for students who spend several hours online each day.

#### Research Question 5. Is there a significant relationship between sleep duration and mental health scores?

```python
import pandas as pd
import plotly.express as px

# Create Sleep Groups
df["Sleep_Group"] = pd.cut(
    df["Sleep_Hours_Per_Night"],
    bins=[0, 5, 7, 10],
    labels=["Low Sleep", "Moderate Sleep", "Adequate Sleep"]
)

# Create Mental Health Groups
df["Mental_Group"] = pd.cut(
    df["Mental_Health_Score"],
    bins=[0, 4, 7, 10],
    labels=["Poor", "Average", "Good"]
)

# Sunburst Chart
fig = px.sunburst(
    df,
    path=["Sleep_Group", "Mental_Group"],
    values="Avg_Daily_Usage_Hours",
    color="Mental_Health_Score",
    color_continuous_scale="Viridis",
    title="Relationship Between Sleep Duration and Mental Health",
    template="plotly_dark"
)

fig.update_layout(title_x=0.5)

fig.show()
```

# 📊 Analysis

#### The Sunburst chart presents a hierarchical view of the relationship between Sleep Duration and Mental Health Scores. The inner ring represents different sleep duration groups (Low, Moderate, and Adequate Sleep), while the outer ring divides each sleep group into Poor, Average, and Good Mental Health categories. The size of each segment is based on the average daily social media usage, and the color intensity reflects the mental health score.

# 💡 Insights
- The chart clearly shows how students are distributed across different sleep duration and mental health categories.
- Students with adequate sleep are generally associated with better mental health categories.
- Students with low sleep duration are more likely to appear in lower mental health categories.
- The hierarchical structure makes it easy to compare mental health outcomes within each sleep group.
- The visualization highlights the connection between healthy sleep habits and overall mental well-being.

#### Research Question 10. What patterns and trends can be identified from the dataset regarding social media usage and overall health?
